# Phi-3

The first step is to load our model onto the GPU for faster inference. Note that we load the model and tokenizer separately (although that isn't always necessary).

当你使用 LLM 时，需要加载两个模型：

- 生成模型本身
- 其底层的分词器（tokenizer）

```python
from transformers import AutoModelForCausalLM, AutoTokenizer

# Load model and tokenizer，使用 transformers 来加载分词器和模型
model = AutoModelForCausalLM.from_pretrained(
    "microsoft/Phi-3-mini-4k-instruct",  # 模型的主路径
    # device_map="cuda",
    device_map=None,  # 不支持GPU
    torch_dtype="auto",
    trust_remote_code=False,
)

tokenizer = AutoTokenizer.from_pretrained("microsoft/Phi-3-mini-4k-instruct")
# 运行上述代码将开始下载模型
# Windows模型下载地址：C:\Users\admin\.cache\huggingface\hub\models--microsoft--Phi-3-mini-4k-instruct
```

Although we can now use the model and tokenizer directly, it's much easier to wrap it in a `pipeline` object:

```python
from transformers import pipeline

# Create a pipeline
# 它将模型、分词器和文本生成过程封装成一个单一的函数
generator = pipeline(
    "text-generation",
    model=model,  # 模型
    tokenizer=tokenizer,  # 分词器
    return_full_text=False,  # 只返回模型的输出结果，而不包含提示词
    max_new_tokens=500,  # 允许模型生成的最大词元数
    do_sample=False  # 决定模型是否使用采样策略来选择下一个词元。将其设置为 False 时，模型将始终选择概率最高的下一个词元
)
```

Finally, we create our prompt as a user and give it to the model:

```python
# The prompt (user input / query)
messages = [
    {"role": "user", "content": "Create a funny joke about chickens."}
]

# Generate output
output = generator(messages)
print(output[0]["generated_text"])
```
